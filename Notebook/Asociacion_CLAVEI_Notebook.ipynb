------ Preparar los datos -------------

```python
from mlxtend.preprocessing import TransactionEncoder
```

Agrupar productos por transacción

```python
transacciones = ventas.groupby('transaccion_id')['item'].apply(list).tolist()
```

Convertir a formato binario

```python
te = TransactionEncoder()
te_array = te.fit(transacciones).transform(transacciones)
```

Crear dataframe final

```python
basket = pd.DataFrame(te_array, columns=te.columns_)
basket.head()
```

Aplicar el algoritmo Apriori o una técnica
equivalente.

```python
from mlxtend.frequent_patterns import apriori
```

Obtener conjuntos frecuentes

```python
frecuentes = apriori(
    basket,
    min_support=0.05,
    use_colnames=True
)

frecuentes.head()
```

este algoritmo Apriori identifica combinaciones frecuentes de productos.

--- Generar reglas de asociación usando soporte, confianza y lift. ---

```python
from mlxtend.frequent_patterns import association_rules
```

---  Generar reglas

```python
reglas = association_rules(
    frecuentes,
    metric='confidence',
    min_threshold=0.30
)
```

Ordenar reglas

```python
reglas = reglas.sort_values(
    by=['lift', 'confidence'],
    ascending=False
)
```

--- Mostrar columnas principales

```python
reglas[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```

**-- Mostrar las 10 reglas más relevantes. --**

```python
reglas_top10 = reglas[
    ['antecedents', 'consequents', 'support', 'confidence', 'lift']
].head(10)

print(reglas_top10)
```
